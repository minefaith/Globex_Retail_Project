## Data Cleaning and Transformation

```python
#drop missing values
globex_df.dropna(subset=['Customer_ID', 'Order_ID'], inplace=True)
```

```python
globex_df.info()
```

```python
# Convert Order_date from string to datetime object
globex_df['Order_Date'] = pd.to_datetime(globex_df ['Order_Date'])
```

```python
globex_df.info()
```

```python
# customer_df = globex_df[['Customer_ID', 'Customer_Location']].copy_duplicates().reset_index(drop=True)
# Customer_df.head()
```

```python
globex_df.columns
```

```python
customers_df = globex_df[['Customer_ID', 'Customer_Location']].copy().drop_duplicates().reset_index(drop=True)
customers_df.head()
```

```python
Orders_df = globex_df[['Order_Date', 'Order_ID']].copy().drop_duplicates().reset_index(drop=True)
Orders_df.head()
```

```python
globex_df.columns
```

```python
products_df = globex_df[['Product_Category', 'Product_Sub_Category']].copy().drop_duplicates().reset_index(drop=True)
products_df.head()
```

```python
price_df = globex_df[['Price', 'Discount', 'Revenue']].copy().drop_duplicates().reset_index(drop=True)
price_df.head()
```

```python
Quantity_df = globex_df[['Quantity']].copy().drop_duplicates().reset_index(drop=True)
Quantity_df.head()
```

```python
customers_df.to_csv(r'dataset/cleaneddata/customers.csv', index=false)
orders_df.to_csv(r'dataset/cleaneddata/orders.csv', index=false)
products_df.to_csv(r'dataset/cleaneddata/products.csv', index=false)
price_df.to_csv(r'dataset/cleaneddata/price.csv', index=false)
Quantity_df.to_csv(r'dataset/cleaneddata/Quantity.csv', index=false)
```
